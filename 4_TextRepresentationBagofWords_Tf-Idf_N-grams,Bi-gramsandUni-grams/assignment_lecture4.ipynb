```python
# lowercase
df['review'] = df['review'].str.lower()
```

```python
# remove html tags
import re
def remove_html_tags(text):
    pattern = re.compile('<.*?>') # The RegEx pattern to find (e.g., <.*?> for HTML tags).
    return pattern.sub(r'', text) # re.sub(pattern, replacement, string)
df['review'] = df['review'].apply(remove_html_tags)
```

```python
df["review"][1]
```

```python
# remove urls
def remove_url(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')  # A common pattern to match URLs
    return pattern.sub(r'', text)

df["review"]=df["review"].apply(remove_url)
```

```python
df["review"][1]
```

```python
# remove punctuations
import string
exclude = string.punctuation
def remove_punc1(text):
    return text.translate(str.maketrans('', '', exclude))

df["review"]=df["review"].apply(remove_punc1)
df["review"][1]
```

```python
# Additional preprocessing - stemming/lemmatization and stopword removal
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download required NLTK data
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

# Remove stopwords
stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    words = text.split()
    return ' '.join([word for word in words if word not in stop_words])

df['review'] = df['review'].apply(remove_stopwords)

# Lemmatization (better than stemming for understanding)
lemmatizer = WordNetLemmatizer()

def lemmatize_text(text):
    words = text.split()
    return ' '.join([lemmatizer.lemmatize(word) for word in words])

df['review'] = df['review'].apply(lemmatize_text)

# Now let's check the cleaned review
print(df["review"][1])
```

```python
# First, install spacy
# !pip install spacy
# !python -m spacy download en_core_web_sm
import spacy

# Load the English model
nlp = spacy.load("en_core_web_sm")


def get_tokens_spacy(text):
    """Tokenize text using spaCy"""
    doc = nlp(text)
    # Return tokens that are not punctuation and not spaces
    tokens = [token.text for token in doc if not token.is_punct and not token.is_space]
    return tokens

# Tokenize all reviews and count
all_tokens = []
for review in df['review']:
    tokens = get_tokens_spacy(review)
    all_tokens.extend(tokens)

total_words = len(all_tokens)
unique_words = set(all_tokens)

print("="*60)
print("PROBLEM 2: WORD COUNT AND VOCABULARY")
print("="*60)
print(f"Total number of words: {total_words}")
print(f"Total unique words (vocabulary): {len(unique_words)}")
print(f"Top 10 most common words: {Counter(all_tokens).most_common(10)}")
```

```python
# Problem 3 - Apply One Hot Encoding
from sklearn.preprocessing import OneHotEncoder
import pandas as pd
import numpy as np

# First, we need to prepare the data for one-hot encoding
# One-hot encoding is typically used for categorical variables
# But for text, we need to create a vocabulary first

# Get all unique words from the corpus
all_words = sorted(set(all_tokens))
print(f"Vocabulary size for one-hot encoding: {len(all_words)}")

# Create a function to one-hot encode a review
def one_hot_encode_review(review, vocabulary):
    """Convert a review to one-hot encoded vector"""
    tokens = get_tokens_spacy(review)
    # Create a dictionary with all words as keys
    one_hot = {word: 0 for word in vocabulary}
    # Set 1 for words present in the review
    for token in set(tokens):  # Use set to avoid duplicates
        if token in one_hot:
            one_hot[token] = 1
    return list(one_hot.values())

# Encode first few reviews (for demonstration)
sample_reviews = df['review'].head(5)
vocabulary_subset = all_words[:100]  # Using subset for demonstration

print("\n" + "="*60)
print("PROBLEM 3: ONE HOT ENCODING")
print("="*60)
print(f"Sample vocabulary (first 10 words): {vocabulary_subset[:10]}")
print(f"Length of one-hot vector: {len(vocabulary_subset)}")

# Encode first review
first_review_encoded = one_hot_encode_review(df['review'].iloc[0], vocabulary_subset)
print(f"\nFirst review one-hot encoding (showing first 20 elements):")
print(first_review_encoded[:20])
print(f"Number of ones in encoding: {sum(first_review_encoded)}")
```

```python
# Problem 4 - Apply Bag of Words
from sklearn.feature_extraction.text import CountVectorizer

print("\n" + "="*60)
print("PROBLEM 4: BAG OF WORDS")
print("="*60)

# Create Bag of Words model
vectorizer_bow = CountVectorizer()
bow_matrix = vectorizer_bow.fit_transform(df['review'])

# Get vocabulary and word frequencies
vocabulary_bow = vectorizer_bow.get_feature_names_out()
word_counts = bow_matrix.toarray().sum(axis=0)

# Create a DataFrame for better visualization
bow_df = pd.DataFrame({
    'word': vocabulary_bow,
    'frequency': word_counts
}).sort_values('frequency', ascending=False)

print(f"Vocabulary size (BoW): {len(vocabulary_bow)}")
print(f"Shape of BoW matrix: {bow_matrix.shape}")
print("\nTop 10 most frequent words:")
print(bow_df.head(10))
print(f"\nWord frequencies for each word:")
for i in range(min(10, len(bow_df))):
    print(f"  {bow_df.iloc[i]['word']}: {int(bow_df.iloc[i]['frequency'])} occurrences")
```

```python
# Problem 5 - Apply Bag of Bi-grams and Tri-grams
print("\n" + "="*60)
print("PROBLEM 5: BAG OF N-GRAMS")
print("="*60)

# Bi-grams
vectorizer_bigram = CountVectorizer(ngram_range=(2, 2))
bigram_matrix = vectorizer_bigram.fit_transform(df['review'])
vocabulary_bigram = vectorizer_bigram.get_feature_names_out()

# Tri-grams
vectorizer_trigram = CountVectorizer(ngram_range=(3, 3))
trigram_matrix = vectorizer_trigram.fit_transform(df['review'])
vocabulary_trigram = vectorizer_trigram.get_feature_names_out()

print(f"Vocabulary size (unigrams): {len(vocabulary_bow)}")
print(f"Vocabulary size (bigrams): {len(vocabulary_bigram)}")
print(f"Vocabulary size (trigrams): {len(vocabulary_trigram)}")

print(f"\nShape of bigram matrix: {bigram_matrix.shape}")
print(f"Shape of trigram matrix: {trigram_matrix.shape}")

# Show some example n-grams
print("\nExample bigrams (first 10):")
for i, bigram in enumerate(vocabulary_bigram[:10]):
    print(f"  {i+1}. {bigram}")

print("\nExample trigrams (first 10):")
for i, trigram in enumerate(vocabulary_trigram[:10]):
    print(f"  {i+1}. {trigram}")

# Observations about dimensionality
print("\n" + "-"*40)
print("OBSERVATIONS ABOUT DIMENSIONALITY:")
print("-"*40)
print(f"1. Unigrams: {len(vocabulary_bow)} dimensions")
print(f"2. Bigrams: {len(vocabulary_bigram)} dimensions")
print(f"3. Trigrams: {len(vocabulary_trigram)} dimensions")
print("\nKey observations:")
print("- As n increases, the vocabulary size decreases significantly")
print("- Bigrams capture more contextual information than unigrams")
print("- Trigrams are more specific and less frequent")
print("- Higher n-grams lead to sparser representations")
```

```python
# Problem 6 - Apply TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer
import math

print("\n" + "="*60)
print("PROBLEM 6: TF-IDF")
print("="*60)

# Create TF-IDF model
vectorizer_tfidf = TfidfVectorizer()
tfidf_matrix = vectorizer_tfidf.fit_transform(df['review'])

# Get vocabulary and IDF scores
vocabulary_tfidf = vectorizer_tfidf.get_feature_names_out()
idf_scores = vectorizer_tfidf.idf_

# Create a DataFrame for better visualization
tfidf_df = pd.DataFrame({
    'word': vocabulary_tfidf,
    'idf_score': idf_scores
}).sort_values('idf_score', ascending=True)  # Low IDF = common words

print(f"Vocabulary size (TF-IDF): {len(vocabulary_tfidf)}")
print(f"Shape of TF-IDF matrix: {tfidf_matrix.shape}")

# Show IDF scores
print("\nTop 10 words with LOWEST IDF scores (most common words):")
for i in range(min(10, len(tfidf_df))):
    print(f"  {tfidf_df.iloc[i]['word']}: {tfidf_df.iloc[i]['idf_score']:.4f}")

print("\nTop 10 words with HIGHEST IDF scores (most unique/rare words):")
tfidf_df_desc = tfidf_df.sort_values('idf_score', ascending=False)
for i in range(min(10, len(tfidf_df_desc))):
    print(f"  {tfidf_df_desc.iloc[i]['word']}: {tfidf_df_desc.iloc[i]['idf_score']:.4f}")

print(f"\nVocabulary (all words):")
print(tfidf_df['word'].tolist())

# Explanation of IDF
print("\n" + "-"*40)
print("IDF SCORE EXPLANATION:")
print("-"*40)
print("IDF (Inverse Document Frequency) measures how rare or common a word is:")
print("- Lower IDF → Word appears in many documents (common)")
print("- Higher IDF → Word appears in few documents (rare/unique)")
print("- IDF = log(Total documents / Number of documents containing the word)")
print(f"- Total documents: {len(df)}")
```

### Complete code

```python
# ============================================
# CELL 1: Import Libraries
# ============================================
import pandas as pd
import numpy as np
import sklearn
import re
import string
from collections import Counter

# ============================================
# CELL 2: Load Data
# ============================================
df = pd.read_csv("IMDB Dataset.csv", nrows=20)

# ============================================
# CELL 3: PROBLEM 1 - Preprocessing
# ============================================
print("\n" + "="*60)
print("PROBLEM 1: TEXT PREPROCESSING")
print("="*60)

# 1. Lowercase
df['review'] = df['review'].str.lower()

# 2. Remove HTML tags
def remove_html_tags(text):
    pattern = re.compile('<.*?>')
    return pattern.sub(r'', text)
df['review'] = df['review'].apply(remove_html_tags)

# 3. Remove URLs
def remove_url(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'', text)
df['review'] = df['review'].apply(remove_url)

# 4. Remove punctuation
def remove_punc(text):
    return text.translate(str.maketrans('', '', string.punctuation))
df['review'] = df['review'].apply(remove_punc)

# 5. Remove stopwords
import nltk
nltk.download('stopwords', quiet=True)
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    words = text.split()
    return ' '.join([word for word in words if word not in stop_words])
df['review'] = df['review'].apply(remove_stopwords)

# 6. Lemmatization
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)
lemmatizer = WordNetLemmatizer()

def lemmatize_text(text):
    words = text.split()
    return ' '.join([lemmatizer.lemmatize(word) for word in words])
df['review'] = df['review'].apply(lemmatize_text)

print("✓ Preprocessing complete!")
print(f"Sample preprocessed review:\n{df['review'][1][:200]}...")

# ============================================
# CELL 4: PROBLEM 2 - Word Count and Vocabulary
# ============================================
import spacy
nlp = spacy.load("en_core_web_sm")

def get_tokens(text):
    doc = nlp(text)
    return [token.text for token in doc if not token.is_punct and not token.is_space]

all_tokens = []
for review in df['review']:
    tokens = get_tokens(review)
    all_tokens.extend(tokens)

print("\n" + "="*60)
print("PROBLEM 2: WORD COUNT AND VOCABULARY")
print("="*60)
print(f"Total number of words: {len(all_tokens)}")
print(f"Total unique words: {len(set(all_tokens))}")
print(f"Top 10 most common words: {Counter(all_tokens).most_common(10)}")

# ============================================
# CELL 5: PROBLEM 3 - One Hot Encoding
# ============================================
from sklearn.preprocessing import OneHotEncoder

print("\n" + "="*60)
print("PROBLEM 3: ONE HOT ENCODING")
print("="*60)

vocabulary = sorted(set(all_tokens))
print(f"Vocabulary size: {len(vocabulary)}")

def one_hot_encode(text, vocab):
    tokens = get_tokens(text)
    encoding = [1 if word in set(tokens) else 0 for word in vocab]
    return encoding

# Demonstrate on first review
first_review_encoded = one_hot_encode(df['review'].iloc[0], vocabulary[:20])  # Using subset
print(f"Sample one-hot encoding (first 20 words): {first_review_encoded}")
print(f"Words present: {sum(first_review_encoded)} out of 20")

# ============================================
# CELL 6: PROBLEM 4 - Bag of Words
# ============================================
from sklearn.feature_extraction.text import CountVectorizer

print("\n" + "="*60)
print("PROBLEM 4: BAG OF WORDS")
print("="*60)

vectorizer_bow = CountVectorizer()
bow_matrix = vectorizer_bow.fit_transform(df['review'])
vocabulary_bow = vectorizer_bow.get_feature_names_out()
word_counts = bow_matrix.toarray().sum(axis=0)

bow_df = pd.DataFrame({
    'word': vocabulary_bow,
    'frequency': word_counts
}).sort_values('frequency', ascending=False)

print(f"Vocabulary size: {len(vocabulary_bow)}")
print(f"Matrix shape: {bow_matrix.shape}")
print("\nTop 10 most frequent words:")
print(bow_df.head(10))

# ============================================
# CELL 7: PROBLEM 5 - Bi-grams and Tri-grams
# ============================================
print("\n" + "="*60)
print("PROBLEM 5: BAG OF N-GRAMS")
print("="*60)

# Bi-grams
vectorizer_bigram = CountVectorizer(ngram_range=(2, 2))
bigram_matrix = vectorizer_bigram.fit_transform(df['review'])
vocabulary_bigram = vectorizer_bigram.get_feature_names_out()

# Tri-grams
vectorizer_trigram = CountVectorizer(ngram_range=(3, 3))
trigram_matrix = vectorizer_trigram.fit_transform(df['review'])
vocabulary_trigram = vectorizer_trigram.get_feature_names_out()

print(f"Unigram vocabulary: {len(vocabulary_bow)}")
print(f"Bigram vocabulary: {len(vocabulary_bigram)}")
print(f"Trigram vocabulary: {len(vocabulary_trigram)}")

print(f"\nSample bigrams: {vocabulary_bigram[:10].tolist()}")
print(f"Sample trigrams: {vocabulary_trigram[:10].tolist()}")

print("\n📊 OBSERVATIONS ON DIMENSIONALITY:")
print("-"*40)
print("As n increases, vocabulary size decreases significantly")
print("Higher n-grams capture more context but create sparser matrices")
print("Bigrams add 2x more features; trigrams add even more specialization")

# ============================================
# CELL 8: PROBLEM 6 - TF-IDF
# ============================================
from sklearn.feature_extraction.text import TfidfVectorizer

print("\n" + "="*60)
print("PROBLEM 6: TF-IDF")
print("="*60)

vectorizer_tfidf = TfidfVectorizer()
tfidf_matrix = vectorizer_tfidf.fit_transform(df['review'])

vocabulary_tfidf = vectorizer_tfidf.get_feature_names_out()
idf_scores = vectorizer_tfidf.idf_

tfidf_df = pd.DataFrame({
    'word': vocabulary_tfidf,
    'idf_score': idf_scores
}).sort_values('idf_score', ascending=True)

print(f"Vocabulary size: {len(vocabulary_tfidf)}")
print(f"Matrix shape: {tfidf_matrix.shape}")

print("\nWords with LOWEST IDF (most common):")
for i in range(min(5, len(tfidf_df))):
    print(f"  {tfidf_df.iloc[i]['word']}: {tfidf_df.iloc[i]['idf_score']:.4f}")

print("\nWords with HIGHEST IDF (most rare/unique):")
tfidf_df_desc = tfidf_df.sort_values('idf_score', ascending=False)
for i in range(min(5, len(tfidf_df_desc))):
    print(f"  {tfidf_df_desc.iloc[i]['word']}: {tfidf_df_desc.iloc[i]['idf_score']:.4f}")

print("\nFull vocabulary:")
print(vocabulary_tfidf.tolist())

print("\n📊 EXPLANATION OF IDF:")
print("-"*40)
print("IDF measures word rarity across documents")
print("Lower IDF → Word appears in many documents (common)")
print("Higher IDF → Word appears in few documents (unique)")
```
