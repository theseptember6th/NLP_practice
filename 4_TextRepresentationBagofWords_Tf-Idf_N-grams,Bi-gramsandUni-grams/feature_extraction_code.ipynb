## bag of words (n_gram=(1,1)) by default

```python
import numpy as np 
import pandas as pd
```

```python
df=pd.DataFrame(
    {
        "text":["people watch campusx",
            "campusx watch campusx",
            "people write comment",
            "campusx write comment"],
        "output":[1,1,0,0]
    }
)
```

```python
df.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer()
```

```python
bow=cv.fit_transform(df["text"])
```

```python
# vocab
print(cv.vocabulary_)
```

```python
len(cv.vocabulary_)
```

```python
print(bow[0].toarray())
print(bow[1].toarray())
```

### prediction

```python
cv.transform(["campusx watch and write comment of campusx"]).toarray() # automatically ignores other non-vocabulary words
```

## for n grams
- bigram n_gram range=(2,2)

```python
import numpy as np 
import pandas as pd
```

```python
df=pd.DataFrame(
    {
        "text":["people watch campusx",
            "campusx watch campusx",
            "people write comment",
            "campusx write comment"],
        "output":[1,1,0,0]
    }
)
```

```python
df.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(ngram_range=(2,2)) # for bigram, (3,3) for trigrams
```

```python
bow=cv.fit_transform(df["text"])
```

```python
# vocab
print(cv.vocabulary_)
```

```python
len(cv.vocabulary_)
```

```python
print(bow[0].toarray())
print(bow[1].toarray())
```

## prediction

```python
cv.transform(["campusx watch and write comment of campusx"]).toarray() # automatically ignores other non-vocabulary words
```

## TFIDF

```python
import numpy as np
import pandas as pd
```

```python
df=pd.DataFrame(
    {
        "text":["people watch campusx",
            "campusx watch campusx",
            "people write comment",
            "campusx write comment"],
        "output":[1,1,0,0]
    }
)
```

```python
df
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer()
tfidf.fit_transform(df["text"]).toarray()
```

```python
print(tfidf.idf_)
print(tfidf.get_feature_names_out())
```
