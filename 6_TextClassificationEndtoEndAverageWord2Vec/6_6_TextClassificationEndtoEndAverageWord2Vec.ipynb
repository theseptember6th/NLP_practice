

## Topic 1: Introduction to Text Classification

### 1. Introduction

**What is Text Classification?**
Text classification is a type of **supervised machine learning** where the goal is to assign a given piece of text (a word, sentence, or entire document) into a predefined category or class.

**Why is it important?**
- It's one of the most commonly used applications in NLP
- The underlying logic works across many different domains
- The same approach can solve many different problems

**Real-life applications:**
- **Email spam filtering** – Classify emails as spam or not spam
- **Sentiment analysis** – Determine if a movie review or product review is positive or negative
- **Customer support** – Route customer queries to the right team (sales vs. support)
- **Language detection** – Identify which language input text is written in (used in Google Translate)
- **Fake news detection** – Classify news articles as real or fake
- **Toxicity detection** – Identify and remove vulgar or inappropriate content on social media

---

### 2. Detailed Explanation

#### What is Classification in General?

Before understanding text classification, let's understand classification:

- Classification is a type of **supervised machine learning**
- Your job is to assign any input data to a specific **class** or category
- Classification can be done on different types of data:
  - Tabular data
  - Image data (e.g., identifying dogs vs. cats)
  - Text data

#### Types of Text Classification

| Type | Description | Example |
|------|-------------|---------|
| **Binary Classification** | Two classes only | Email spam vs. not spam |
| **Multi-class Classification** | More than two classes, but each input belongs to only one class | News article → Sports, Entertainment, Business, Politics |
| **Multi-label Classification** | Each input can belong to multiple classes simultaneously | A news article can be about Sports, Cricket, and Sachin Tendulkar all at once |

> **Note:** Binary and multi-class are most common. Multi-label is less common but you may encounter it.

---

### 3. Key Points

- Text classification is supervised learning for text data
- Input can be a word, sentence, or complete document
- Three types: binary, multi-class, and multi-label
- Used across many industries and applications
- The logic remains similar across different domains

- Thinking text classification is only for sentiment analysis – it's used for many other tasks too
- Not understanding that the same approach can solve different problems

# Topic 3: Heuristic (Jugaad) Approach & API-based Approach

---

### 1. Introduction

Not every text classification problem requires complex machine learning or deep learning. Sometimes, you can solve problems with **simple rules** (heuristic approach), and other times you can use **existing services** (API-based approach). Both are valid options depending on your situation.

---

### 2. Detailed Explanation

#### Part A: Heuristic / Jugaad Approach

**What is it?**
This is a rule-based approach where you create your own logic without using any machine learning or deep learning algorithms. You manually define rules to classify text.

**When to use it:**
- When you have **no data** to train a model
- When you're working on a very specific problem with rare text (e.g., classified government projects)
- When you need a quick, temporary solution

**Example:**
Suppose you're doing sentiment analysis but have no labeled data. You could create a simple rule:

```
1. Count how many positive words are in the text (e.g., "good", "excellent", "amazing")
2. Count how many negative words are in the text (e.g., "bad", "terrible", "worst")
3. If positive_count > negative_count → classify as POSITIVE
   Else → classify as NEGATIVE
```

> **Historical context:** This approach was commonly used before machine learning and deep learning became widespread. Today, it's rarely used except in data-scarce situations.

**When you might combine datasets:**
Sometimes, if you have multiple small datasets, you can combine them to create a larger dataset for training. This is also considered a heuristic approach.

---

#### Part B: API-based Approach

**What is it?**
This involves using ready-made APIs (Application Programming Interfaces) provided by companies like Google, Amazon, or third-party providers. You send your text to their service, and they return the classification result.

**When to use it:**
- When you don't want to develop the solution yourself
- When your company is willing to pay for the service
- When you need quick results without investing in development

**Example:**
- **Google Cloud Natural Language API** – provides sentiment analysis, entity recognition, etc.
- **AWS Comprehend** – provides text classification and analysis
- **NLP Crowd API** – offers various NLP services including sentiment analysis

**How it works (step-by-step):**
1. You sign up for the API service
2. You send your text to their endpoint
3. The API processes it using their pre-trained models
4. You receive the classification result (e.g., "positive" with 99% confidence)

**Pros of using APIs:**
- Saves development effort and time
- No need to maintain your own models
- Often state-of-the-art performance

**Cons of using APIs:**
- **Costs money** – you pay per request
- Limited customization
- Your data goes to a third party (privacy concerns)

# Topic 5: TF-IDF (Term Frequency-Inverse Document Frequency)

---

### 1. Introduction

**What is TF-IDF?**
TF-IDF (Term Frequency-Inverse Document Frequency) is a **text vectorization technique** that goes beyond simple word counting. Instead of just counting how many times a word appears in a document (like Bag of Words), TF-IDF **weighs** each word based on how **important** it is to that specific document.

**Why is it important?**
- Common words like "the", "is", "and" appear frequently in all documents but carry little meaning
- Rare words that appear in only a few documents are often more **informative**
- TF-IDF helps highlight important words while downplaying common ones

**Real-life use:**
- Information retrieval systems (search engines)
- Text classification
- Document clustering
- Keyword extraction

---

### 2. Detailed Explanation

#### The Problem with Bag of Words

Imagine you have 3 documents:
- Document 1: "The cat sat on the mat"
- Document 2: "The dog sat on the mat"  
- Document 3: "The elephant sat on the mat"

In Bag of Words, words like "the", "on", "sat", "mat" will appear in ALL documents, making them equally important in all vectors. But these words don't help distinguish Document 1 from Document 2. Only "cat", "dog", and "elephant" are actually useful for classification.

**TF-IDF solves this** by giving **higher weight** to words that appear frequently in one document but rarely in others.

#### Step-by-Step Explanation

TF-IDF is composed of two parts:

**Part 1: Term Frequency (TF)**
- Measures how frequently a word appears in a specific document
- Formula: `TF(t, d) = (Number of times term t appears in document d) / (Total number of terms in document d)`

**Part 2: Inverse Document Frequency (IDF)**
- Measures how common or rare a word is across ALL documents
- Formula: `IDF(t) = log(Total number of documents / Number of documents containing term t)`

**Part 3: TF-IDF Score**
- The final score is: `TF-IDF(t, d) = TF(t, d) × IDF(t)`

#### Example to Understand

Consider 1000 documents. The word "the" appears in all 1000 documents:
- TF for a specific document: Let's say 0.05 (appears 5 out of 100 words)
- IDF: log(1000/1000) = log(1) = 0
- TF-IDF = 0.05 × 0 = 0 → "the" gets NO weight

Now consider the word "machine" appearing in only 10 out of 1000 documents:
- TF for a document: 0.01
- IDF: log(1000/10) = log(100) = 2
- TF-IDF = 0.01 × 2 = 0.02 → "machine" gets higher weight

Words like "machine" that appear rarely across documents but frequently within a document get the **highest** TF-IDF scores.

#### Using TF-IDF for Text Classification

In the transcript, the instructor demonstrated TF-IDF for the same IMDb dataset:

| Vectorization Method | Algorithm | Accuracy |
|----------------------|-----------|----------|
| Bag of Words | Random Forest | ~84.06% |
| **TF-IDF** | **Random Forest** | **~84.8%** |

> **Result:** TF-IDF gave a **slight improvement** over Bag of Words (84.8% vs 84.06%).

---

### 3. Key Points

- **TF-IDF** = Term Frequency × Inverse Document Frequency
- **Term Frequency** = how often a word appears in a document
- **Inverse Document Frequency** = how rare a word is across all documents
- **Common words** (stopwords) get low IDF scores → low weight
- **Rare but meaningful words** get high IDF scores → high weight
- TF-IDF often gives **better results** than plain Bag of Words
- You can further improve by **tuning parameters** (e.g., n-gram range)
- TF-IDF is used both in **information retrieval** (search) and **classification**

---

# Topic 6: Average Word2Vec for Text Classification

---

### 1. Introduction

**What is Average Word2Vec?**
Word2Vec is a technique that converts each **word** into a **dense vector** (typically 300 dimensions) that captures semantic meaning. But for text classification, we need a vector for an **entire document or sentence**, not just individual words. **Average Word2Vec** solves this by averaging all the word vectors in a document to create a single document vector.

**Why is it important?**
- Word2Vec captures **semantic relationships** between words (e.g., "king" and "queen" are close in vector space)
- It handles words not seen during training better than Bag of Words
- It gives more meaningful representations than sparse Bag of Words vectors
- Often gives **better results** than BoW/TF-IDF when you have enough data

**Real-life use:**
- Text classification
- Document similarity
- Search engines
- Recommendation systems

---

### 2. Detailed Explanation

#### Part A: Word2Vec Overview (Quick Recap)

**What Word2Vec does:**
- Converts each word into a **dense vector** (e.g., 100 or 300 numbers)
- Words with similar meanings have **similar vectors** (close in vector space)
- Example: "happy" and "joyful" will have vectors close to each other
- Example: "cat" and "dog" will be closer than "cat" and "car"

**How it's trained:**
- You either:
  1. Use a **pre-trained** Word2Vec model (e.g., Google News vectors)
  2. **Train your own** Word2Vec on your specific dataset

#### Part B: The Problem – Document Level Representation

**The challenge:**
- Word2Vec gives word-level vectors (300 numbers per word)
- But we need **document-level** vectors for classification
- A movie review has hundreds of words – we need one vector per review

**The solution: Average Word2Vec**

#### Step-by-Step Explanation of Average Word2Vec

**Step 1:** Get the word vector for each word in the document

Suppose we have the sentence: "I love campus"
- "I" → [0.2, -0.5, 0.1, ...] (300 numbers)
- "love" → [0.8, 0.3, -0.2, ...] (300 numbers)
- "campus" → [-0.1, 0.6, 0.4, ...] (300 numbers)

**Step 2:** Add all the word vectors together (vector addition)

For each dimension, add the corresponding values:
- Dimension 1: 0.2 + 0.8 + (-0.1) = 0.9
- Dimension 2: (-0.5) + 0.3 + 0.6 = 0.4
- Dimension 3: 0.1 + (-0.2) + 0.4 = 0.3
- ...and so on for all 300 dimensions

**Step 3:** Divide by the number of words (average)

- Dimension 1: 0.9 / 3 = 0.3
- Dimension 2: 0.4 / 3 = 0.13
- Dimension 3: 0.3 / 3 = 0.1

**Step 4:** The result is a single 300-dimensional vector representing the entire sentence

> **Key insight:** The averaged vector captures the **average meaning** of all words in the document.

#### Part C: Using Average Word2Vec for Classification (as shown in the transcript)

**Process:**

1. **Train Word2Vec on your data** (or load pre-trained)
   - The instructor used only the first 1000 reviews for demonstration
   - Trained word2vec on these reviews
   - Each word becomes a 100-dimensional vector

2. **Average all word vectors** in each review to get one document vector
   - Review 1 → 100-dimensional vector
   - Review 2 → 100-dimensional vector
   - All reviews converted to 100-dimensional vectors

3. **Apply machine learning** on the document vectors
   - Used **Random Forest** classifier

4. **Evaluate** the model

**Result from the transcript:**
- Accuracy with Average Word2Vec: **~70%**
- This was lower than Bag of Words (84%) and TF-IDF (84.8%)

**Why was the result lower?**
1. **Small training size** – Only used first 1000 reviews to train word2vec
   - With more data, Word2Vec embeddings improve
2. **Vocabulary mismatch** – Word2Vec may not capture all words well with limited data
3. **Averaging loses information** – The order and context of words is lost
   - "I love this" and "I hate this" could have similar averages!

### 3. Key Points

- **Average Word2Vec** = average all word vectors in a document → document vector
- Word2Vec captures **semantic meaning** (words with similar meanings have similar vectors)
- You can **train your own** Word2Vec on your dataset OR use **pre-trained** vectors
- **Pre-trained rule of thumb:** At least 40% of words in your vocabulary should match the pre-trained model's vocabulary
- **More data = better embeddings** – The 70% result was due to limited data
- Averages **lose word order information** – "good bad" and "bad good" could have similar vectors
- Despite the lower result in this example, Word2Vec often gives **better results** than BoW/TF-IDF with sufficient data

### 4. Syntax/Structure (Code Concepts)

```python
# Step 1: Train Word2Vec
from gensim.models import Word2Vec

# Tokenize sentences into words
sentences = [review.split() for review in X_train]

# Train Word2Vec model
w2v_model = Word2Vec(sentences, vector_size=100, window=5, min_count=1)

# Step 2: Function to average word vectors
def average_word2vec(review, model, vector_size=100):
    vectors = [model.wv[word] for word in review.split() if word in model.wv]
    if len(vectors) > 0:
        # Average all word vectors
        return np.mean(vectors, axis=0)
    else:
        # Return zero vector if no words found
        return np.zeros(vector_size)

# Step 3: Convert all reviews to document vectors
X_vectors = np.array([average_word2vec(review, w2v_model) for review in X_train])

# Step 4: Train classifier (Random Forest)
model = RandomForestClassifier()
model.fit(X_vectors, y_train)

# Step 5: Evaluate
predictions = model.predict(X_test_vectors)
accuracy = accuracy_score(y_test, predictions)
```

### 6. Common Mistakes

- **Using too little data for Word2Vec** – You need sufficient data to train good embeddings
- **Not checking vocabulary overlap** – If using pre-trained vectors, ensure at least 40% of your words are in the pre-trained vocabulary

- **Losing word order information** – Averaging discards the sequence of words. For some tasks, order matters.

### 2. Detailed Explanation – 5 Key Tips

#### Tip 1: Use Handcrafted Features Along with Vectorization

**What are handcrafted features?**
These are **domain-specific features** that you create manually based on your understanding of the data.

**Examples:**
- For sentiment analysis: count of exclamation marks (!), number of positive/negative words
- For spam detection: presence of links, number of uppercase words, use of "$" symbols
- For news classification: presence of specific keywords (e.g., "election" → politics)

**Why combine with machine learning?**
- Machine learning (BoW, TF-IDF) alone might not capture everything
- Handcrafted features add domain expertise
- **Combination** of both often performs better than either alone

**Example:**
```
Feature 1: TF-IDF vectors (from the text)
Feature 2: Number of positive words in review (handcrafted)
Feature 3: Number of exclamation marks (handcrafted)
→ Combine all features and train classifier
```

---

#### Tip 2: Start with Machine Learning, NOT Deep Learning

**The advice:**
Don't jump straight to deep learning. Start with traditional machine learning first.

**Reasons:**
1. **Interpretability** – ML models are easier to understand and debug
   - You can see which features are important
   - You can explain decisions to stakeholders
2. **Understand your data better** – Working with ML forces you to explore and understand the problem
3. **Deep learning is "black box"** – You lose visibility into what's happening internally
4. **ML models can outperform DL** – Sometimes, simpler models give better results
5. **Easier deployment** – ML models are smaller, faster, and easier to deploy

**Recommendation:**
Machine Learning → Deep Learning (only if needed)

**When to move to deep learning?**
- When ML models have plateaued
- When you have LARGE amounts of data (millions of samples)
- When you need state-of-the-art results

---

#### Tip 3: Handle Imbalanced Datasets

**What is an imbalanced dataset?**
When one class has significantly more samples than another.

**Example:**
- 80% of reviews are positive, 20% are negative
- If you predict "positive" for everything, you get 80% accuracy – but your model is useless!

**Why it's a problem:**
- Models become biased toward the majority class
- Accuracy becomes a misleading metric
- The model fails to learn the minority class properly

**How to handle it:**

| Technique | Description |
|-----------|-------------|
| **Resampling** | Oversample minority class OR undersample majority class |
| **Use different metrics** | Precision, Recall, F1-Score (not just accuracy) |
| **Class weights** | Give higher weight to minority class during training |
| **Synthetic data** | Generate synthetic samples (e.g., SMOTE) |

**Key message:** Always check class balance before training!

---

#### Tip 4: Evaluate and Interpret Your Models

**Don't just look at accuracy!**

| Metric | What it Measures | When to Use |
|--------|------------------|-------------|
| **Accuracy** | Overall correct predictions | Balanced datasets |
| **Precision** | Of positive predictions, how many were correct | When false positives are costly |
| **Recall** | Of actual positives, how many were caught | When false negatives are costly |
| **F1-Score** | Harmonic mean of precision & recall | Imbalanced datasets |
| **Confusion Matrix** | Shows all prediction outcomes | Always use this |

**How to interpret your model:**
- Look at **confusion matrix** to see where the model is making mistakes
- Analyze **misclassified examples** – what patterns do they share?
- Use feature importance (for Random Forest, etc.) to understand what the model "learned"
