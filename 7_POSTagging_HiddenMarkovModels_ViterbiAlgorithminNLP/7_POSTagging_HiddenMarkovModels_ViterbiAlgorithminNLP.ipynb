## Topic 1: Introduction to POS Tagging & Its Applications

### 1. Introduction

**What is POS Tagging?**
POS (Part-of-Speech) Tagging is the process of assigning a grammatical category (like noun, verb, adjective, etc.) to **each word** in a given sentence.

- **Example:** In the sentence *"Nitesh loves camping"* — "Nitesh" gets the tag **Noun (NN)**, "loves" gets **Verb (VB)**, and "camping" gets **Noun (NN)**.

**Why is it important?**
POS tagging is a **preprocessing step** in most Natural Language Processing (NLP) pipelines. It helps machines understand the syntactic structure of a sentence, which is essential for any advanced language task.

**Real-life uses:**
- **Information Retrieval (NER):** Extracting names, dates, and locations (e.g., "Rahul" → Person, "March 30" → Date).
- **Question Answering Systems:** Chatbots and virtual assistants use POS to parse user queries.
- **Word Sense Disambiguation:** Distinguishing the meaning of a word based on context (e.g., "left" in "I left the room" vs. "the left side").
- **Sentiment Analysis & Text Summarization:** Understanding sentence structure improves analysis.

---

### 2. Detailed Explanation

#### What is a "Part of Speech"?
In grammar, a part of speech is a category of words that have similar grammatical properties. You may remember:
- **Noun (NN):** Person, place, or thing (e.g., *Nitesh, Google*).
- **Verb (VB):** Action or state (e.g., *go, read*).
- **Adjective (ADJ):** Describes a noun (e.g., *big, fast*).
- **Preposition (IN):** Shows relationship (e.g., *in, on, at*).
- **Modal (MD):** Auxiliary verbs (e.g., *will, can, must*).

#### The Core Idea
When you provide a sentence as input, POS tagging analyzes it **word-by-word** and tags each word with its appropriate POS.

#### Why is this a Preprocessing Step?
In any NLP pipeline, raw text is first cleaned, tokenized (split into words), and then POS tagged before feeding it into advanced models. Without this step, machines cannot determine the role of each word in a sentence.

#### Example: Word Sense Disambiguation
Consider the word **"read"**:
- *"I **read** a book on history."* → Here, "read" is in **past tense** (VBD).
- *"I **read** a book every day."* → Here, "read" is in **present tense** (VBP).

POS tagging correctly identifies the tense based on **context**, helping the machine interpret meaning accurately.

---

### 3. Key Points

- POS tagging assigns grammatical labels to every word in a sentence.
- It is a **preprocessing step**, not a final task.
- It is crucial for tasks like NER, question answering, and word sense disambiguation.
- The same word can have different POS tags depending on context.
- The output is a sequence of tags corresponding to the word sequence.

---

### 4. Common Mistakes

- **Assuming POS tagging is the final goal:** It is only a helper step for bigger systems.
- **Confusing "word meaning" with "word tag":** A word like "run" can be a noun ("a run") or a verb ("to run"). The tag depends on sentence context, not the dictionary definition.
- **Thinking all NLP tasks use POS:** While many do, it is not mandatory for every application.

---

### 5. Interview/Exam Questions

**Q1: What is POS tagging and why is it important?**  
**A:** POS tagging assigns grammatical categories to words in a sentence. It is important because it provides syntactic structure needed for advanced NLP tasks like NER, QA systems, and disambiguation.

**Q2: Give an example where POS tagging helps in word sense disambiguation.**  
**A:** In *"I left the room"* and *"the left side of the room"*, POS tagging distinguishes "left" as a verb in the first sentence and as an adjective in the second, resolving ambiguity.

**Q3: Is POS tagging a preprocessing or post-processing step?**  
**A:** It is a preprocessing step in the NLP pipeline.

---

### 6. Revision Notes

- **Definition:** Assigning POS tags (Noun, Verb, etc.) to each word in a sentence.
- **Nature:** Preprocessing step.
- **Key Applications:** NER, QA, disambiguation, sentiment analysis.
- **Context matters:** Same word can have different tags in different sentences.
- **Outcome:** A sequence of tags that helps machines understand sentence structure.

## Topic 2: Hands-on POS Tagging using spaCy (Code, Attributes & Visualization)

### 1. Introduction

**What is spaCy?**
spaCy is a popular, open-source Python library designed specifically for advanced NLP tasks. It is **production-ready** and comes with pre-trained models that can perform POS tagging, tokenization, named entity recognition, and more—right out of the box.

**Why use spaCy for POS tagging?**
- It is **fast** and **efficient**.
- It provides **two levels** of POS tags: coarse-grained (simplified) and fine-grained (detailed).
- It includes a built-in visualizer (`displacy`) to see the tags graphically.
- You don't need to train a model; spaCy's pre-trained model does all the heavy lifting.

**Real-life use:**
If you are building a chatbot, a text summarizer, or any language tool, you can use spaCy to quickly get POS tags without writing complex algorithms yourself.

---

### 2. Detailed Explanation

#### Step-by-step workflow

1. **Install spaCy** (if not already installed).
2. **Import the library** and load the English language model.
3. **Create a document object** from your input sentence.
4. **Access tags** using attributes like `.pos_` (coarse) and `.tag_` (fine-grained).
5. **Use the `spacy.explain()`** function to get the full meaning of a tag.
6. **Visualize** the tags using `displacy`.

---

#### Understanding the Attributes

When you process a text with spaCy, it returns a `Doc` object. Each token (word) in this object has several important attributes:

| Attribute | What it gives | Example |
|-----------|---------------|---------|
| `.text` | The actual word | `"Google"` |
| `.pos_` | Coarse-grained POS tag (short, general) | `"PROPN"` (proper noun) |
| `.tag_` | Fine-grained POS tag (more specific) | `"NNP"` (proper noun, singular) |
| `.pos` / `.tag` | Integer IDs (not used for beginners) | — |

**Coarse vs. Fine-grained:**
- **Coarse (`.pos_`):** Broad category (e.g., `NOUN`, `VERB`, `ADJ`). Good for general understanding.
- **Fine-grained (`.tag_`):** Detailed tag (e.g., `NNP` for proper noun, `VBD` for past tense verb). Good for precise grammar analysis.

**Using `spacy.explain()`**
If you don't understand a tag, you can pass it to `spacy.explain()` to get a plain-English description.

Example:
```python
spacy.explain("PRP")   # Output: "pronoun, personal"
spacy.explain("VBD")   # Output: "verb, past tense"
```

- spaCy does POS tagging automatically behind the scenes.
- You access tags using `.pos_` and `.tag_` on each token.
- Fine-grained tags give more grammatical detail than coarse ones.
- `spacy.explain()` helps you understand what each tag means.

## Topic 3: The Hidden Markov Model (HMM) for POS Tagging – Step-by-Step

### 1. Introduction

**What is a Hidden Markov Model (HMM)?**
A Hidden Markov Model is a statistical model that assumes there is an **underlying (hidden) sequence** of states (in our case, POS tags) that generates the **observed sequence** (the words we see). We don't directly see the tags—they are "hidden"—but we can infer them from the words.

**Why is HMM used for POS Tagging?**
POS tagging is a classic **sequence labeling** problem. For each word in a sentence, we need to assign a tag. The tag of a word often depends on:
- The word itself (how likely is this word to be a noun?).
- The previous tag (what tag is likely to follow another tag?).

HMM captures both these dependencies beautifully using probabilities. It is one of the most fundamental and widely taught approaches for POS tagging.

**Real-life use:**
Before deep learning became popular, HMMs were the go-to method for POS tagging, speech recognition, and even DNA sequencing. Today, they remain an excellent conceptual foundation for understanding sequence models.

---

### 2. Detailed Explanation

#### The Core Idea

We have:
- A sequence of **observed words**: \( W = w_1, w_2, ..., w_n \)
- A sequence of **hidden tags**: \( T = t_1, t_2, ..., t_n \)

We want to find the **most probable tag sequence** \( T^* \) given the words:
\[
T^* = \arg\max_T P(T | W)
\]

Using Bayes' rule, this becomes:
\[
T^* = \arg\max_T P(W | T) \times P(T)
\]

HMM breaks this down into two types of probabilities:

1. **Emission Probabilities** \( P(w_i | t_i) \): Given a tag, what is the probability that it produces a specific word?
   - Example: \( P(\text{"Google"} | \text{NOUN}) \) = high, because "Google" is often a proper noun.

2. **Transition Probabilities** \( P(t_i | t_{i-1}) \): Given the previous tag, what is the probability of the current tag?
   - Example: \( P(\text{VERB} | \text{NOUN}) \) = probability that a verb follows a noun.

Plus, we add a **Start** and **End** tag to handle sentence boundaries.

---

#### Training the HMM (Step-by-step from the transcript)

The transcript walks us through training an HMM on a small dataset. Let's reconstruct that process clearly.

**Step 1: The Training Data**

We have 4 sentences with their POS tags manually assigned:

| Sentence | Tags |
|----------|------|
| Nitesh loves camping | NOUN VERB NOUN |
| Nitesh Google camping | NOUN NOUN NOUN |
| Will Ankita Google camping | MODAL NOUN NOUN NOUN |
| Ankita loves will | NOUN VERB MODAL |

**(Note:** In the transcript, the tags are simplified to only 3 types: **NOUN**, **VERB**, and **MODAL** for clarity.)

**Step 2: Identify Unique Words and Tags**

- Unique words: `{Nitesh, loves, camping, Google, Will, Ankita, will, can}`
- Unique tags: `{NOUN, VERB, MODAL}`
- Plus `START` and `END` for sentence boundaries.

**Step 3: Calculate Emission Probabilities**

We count how many times each word appears under each tag, then divide by the total count of that tag.

| Word | Count with NOUN | Count with VERB | Count with MODAL |
|------|----------------|----------------|------------------|
| Nitesh | 2 | 0 | 0 |
| loves | 0 | 2 | 0 |
| camping | 3 | 0 | 0 |
| Google | 3 | 0 | 0 |
| Will | 0 | 0 | 2 |
| Ankita | 2 | 0 | 0 |
| will | 0 | 0 | 1 |
| can | 0 | 0 | 1 |
| **Total** | **10** | **2** | **4** |

Now, convert to probabilities by dividing each column by its total:

| Word | \( P(\text{word} \mid \text{NOUN}) \) | \( P(\text{word} \mid \text{VERB}) \) | \( P(\text{word} \mid \text{MODAL}) \) |
|------|----------------------------------------|----------------------------------------|------------------------------------------|
| Nitesh | 2/10 = 0.2 | 0 | 0 |
| loves | 0 | 2/2 = 1.0 | 0 |
| camping | 3/10 = 0.3 | 0 | 0 |
| Google | 3/10 = 0.3 | 0 | 0 |
| Will | 0 | 0 | 2/4 = 0.5 |
| Ankita | 2/10 = 0.2 | 0 | 0 |
| will | 0 | 0 | 1/4 = 0.25 |
| can | 0 | 0 | 1/4 = 0.25 |

**Interpretation:**
- If a word is a NOUN, there is a 20% chance it is "Nitesh".
- If a word is a VERB, there is a 100% chance it is "loves" (in this tiny dataset).

---

**Step 4: Calculate Transition Probabilities**

We add START and END to every sentence. Then we count tag transitions.

Example: For sentence 1 `START → NOUN → VERB → NOUN → END`

We create a table of counts: from previous tag → current tag.

| From \ To | START | NOUN | VERB | MODAL | END |
|-----------|-------|------|------|-------|-----|
| START     | 0     | 3    | 0    | 1     | 0   |
| NOUN      | 0     | 0    | 2    | 1     | 5   |
| VERB      | 0     | 2    | 0    | 0     | 0   |
| MODAL     | 0     | 2    | 0    | 0     | 0   |
| END       | 0     | 0    | 0    | 0     | 0   |

Now convert each row to probabilities by dividing by the row total:

| From \ To | START | NOUN | VERB | MODAL | END |
|-----------|-------|------|------|-------|-----|
| START     | 0     | 3/4=0.75 | 0    | 1/4=0.25 | 0   |
| NOUN      | 0     | 0    | 2/8=0.25 | 1/8=0.125 | 5/8=0.625 |
| VERB      | 0     | 2/2=1.0 | 0    | 0    | 0   |
| MODAL     | 0     | 2/2=1.0 | 0    | 0    | 0   |
| END       | 0     | 0    | 0    | 0    | 0   |

**Interpretation:**
- After START, there is a 75% chance the next tag is NOUN, and 25% chance it is MODAL.
- After NOUN, there is a 62.5% chance the sentence ends (END), 25% chance a VERB follows, and 12.5% chance a MODAL follows.
- After VERB, there is a 100% chance a NOUN follows.
- After MODAL, there is a 100% chance a NOUN follows.

---

**Step 5: Build the HMM Graph**

We now have a directed graph with:
- **States:** START, NOUN, VERB, MODAL, END
- **Transition probabilities:** From each state to the next (from the table above).
- **Emission probabilities:** From each tag state to each word (from the emission table).

This graph represents our trained HMM.

---

**Step 6: Using the HMM for Tagging a New Sentence**

Suppose we want to tag the sentence: **"Will Google camping"**

We can now compute the probability of every possible tag sequence and pick the one with the highest overall probability.

For a sequence like:
`START → MODAL → NOUN → NOUN → END`
The probability is:
\[
P(\text{START}) \times P(\text{MODAL}|\text{START}) \times P(\text{Will}|\text{MODAL}) \times P(\text{NOUN}|\text{MODAL}) \times P(\text{Google}|\text{NOUN}) \times P(\text{NOUN}|\text{NOUN}) \times P(\text{camping}|\text{NOUN}) \times P(\text{END}|\text{NOUN})
\]

We multiply all these probabilities together. The sequence with the **maximum product** is chosen as the final tag sequence.

---

### 3. Key Points

- HMM models POS tagging as a **sequence labeling** problem.
- It uses **two probability tables**: Emission (word → tag) and Transition (tag → tag).
- Training involves counting occurrences in labeled data.
- Inference (tagging) involves finding the tag sequence with the highest probability.
- The START and END symbols help capture sentence boundaries.

---

### 4. Common Mistakes

- **Forgetting to add START and END:** This leads to incorrect transition counts at sentence edges.
- **Using raw counts instead of probabilities:** Always normalize counts to probabilities to avoid bias from sentence length.
- **Assuming all tag sequences are equally likely:** Transition probabilities handle this; don't ignore them.
- **Confusing emission and transition:** Emission = word given tag; Transition = tag given previous tag.

---

### 5. Interview/Exam Questions

**Q1: What are the two types of probabilities in an HMM for POS tagging?**  
**A:** Emission probabilities \( P(\text{word} \mid \text{tag}) \) and Transition probabilities \( P(\text{current tag} \mid \text{previous tag}) \).

**Q2: Why do we add START and END tags in HMM?**  
**A:** To model the probability of beginning and ending a sentence, which helps capture grammatical patterns at boundaries.

**Q3: How do you train an HMM for POS tagging?**  
**A:** By counting occurrences of tag-word pairs (emission) and tag-tag pairs (transition) in a labeled corpus, then normalizing to probabilities.

**Q4: What is the goal during inference with an HMM?**  
**A:** To find the tag sequence that maximizes the product of emission and transition probabilities given the observed words.

---

### 6. Revision Notes

- **HMM**: A generative model for sequence labeling.
- **Emission Prob.**: \( P(word | tag) \) – probability a tag produces a word.
- **Transition Prob.**: \( P(current tag | previous tag) \) – probability a tag follows another.
- **Training**: Count and normalize from labeled data.
- **Inference**: Find the most probable tag sequence using all probabilities.
- **START/END**: Added to model sentence beginnings and endings.

## Topic 4: The Brute-Force Approach & Its Computational Problem

### 1. Introduction

**What is the Brute-Force Approach?**
When we have a trained HMM and we want to tag a new sentence, the most straightforward method is to:
1. Generate **every possible combination** of POS tags for the words in the sentence.
2. Calculate the probability of each combination using the emission and transition probabilities.
3. Pick the combination with the **highest probability**.

This is called the **brute-force** or **exhaustive search** approach.

**Why is it important to understand this?**
The brute-force method is conceptually simple and perfectly correct—it will always find the optimal tag sequence. However, it is **computationally infeasible** for real-world sentences. Understanding *why* it fails motivates the need for smarter algorithms like the Viterbi algorithm.

**Real-life use:**
You would never use brute-force in practice. But it serves as a baseline: any efficient algorithm (like Viterbi) must produce exactly the same result as brute-force, just faster.

---

### 2. Detailed Explanation

#### How Brute-Force Works (Step-by-Step)

Let's say we have a sentence with **n** words. Each word can take **t** possible POS tags. For example:
- In our earlier dataset, t = 3 (NOUN, VERB, MODAL).
- If our sentence has 4 words (like "Will Google camping"), the total number of tag combinations is:

\[
\text{Total combinations} = t^n = 3^4 = 81
\]

For each combination (e.g., NOUN-VERB-NOUN-MODAL), we:
1. Multiply the emission probability of each word given its assigned tag.
2. Multiply the transition probability from START to the first tag, between each consecutive pair, and from the last tag to END.
3. Get a single probability score.

After computing all 81 scores, we pick the combination with the highest score.

---

#### The Problem: Exponential Explosion

The number of combinations grows **exponentially** with sentence length:

| Sentence Length (n) | Total Combinations (t=3) | Total Combinations (t=45, typical POS tags) |
|----------------------|--------------------------|---------------------------------------------|
| 1                    | 3                        | 45                                          |
| 2                    | 9                        | 2,025                                       |
| 3                    | 27                       | 91,125                                      |
| 4                    | 81                       | ~4 million                                  |
| 5                    | 243                      | ~184 million                                |
| 10                   | 59,049                   | ~3.4 × 10^16 (astronomically large)         |
| 20                   | 3.4 × 10^9               | ~10^33 (more than atoms in the universe)    |

**Key insight:** In real NLP, we have around **45–50 POS tags** and sentences can be 10–20 words long. The number of combinations becomes **astronomical**—far too many for any computer to evaluate.

---

#### Why Can't We Just Use Brute-Force?

- **Time complexity:** \( O(t^n) \) – exponential time. For a 20-word sentence with 45 tags, even a supercomputer would take longer than the age of the universe.
- **Memory:** Storing all combinations is impossible.
- **Real-time systems:** Chatbots and search engines need answers in milliseconds, not years.

**The core problem:**
We are doing **repeated work**. Many different tag sequences share the same prefix. For example:
- `NOUN-VERB-NOUN-...` and `NOUN-VERB-MODAL-...` both start with `NOUN-VERB`. Brute-force recomputes the probability of that prefix multiple times.

---

#### A Concrete Example from the Transcript

In the transcript, the sentence used was: **"Will Google camping"** (4 words). With 3 tags, total combinations = 81.

The transcript shows that for the **correct** combination:  
`START → MODAL → NOUN → NOUN → END`

The probability was **non-zero** and turned out to be the highest.

But many other combinations gave a **probability of zero** because:
- A transition probability was zero (e.g., `NOUN → MODAL` might have been 0 in the training data).
- An emission probability was zero (e.g., `P("Will" | NOUN)` was 0).

Even with these zeros, the algorithm still had to **check every single combination** to know they were zero. That's wasted computation.

---

### 3. Key Points

- Brute-force generates **all possible tag sequences** and scores each one.
- The number of sequences is \( t^n \) where \( t \) = number of tags, \( n \) = sentence length.
- This grows **exponentially**, making it impractical for sentences longer than 4–5 words.
- The approach is **conceptually simple** but **computationally explosive**.
- Many sequences share prefixes, leading to **redundant calculations**.

---

### 4. Common Mistakes

- **Thinking brute-force is "good enough" for small sentences:** Even a 10-word sentence with 45 tags gives ~3.4 × 10^16 combinations—completely unmanageable.
- **Confusing the number of tags with the number of words:** Both contribute to the explosion, but the exponent is the sentence length, which is the bigger driver.
- **Assuming we can prune zero-probability paths early:** You can, but that's exactly what smarter algorithms do—brute-force does not.

---

### 5. Interview/Exam Questions

**Q1: Why is the brute-force approach not used for POS tagging in practice?**  
**A:** Because the number of tag combinations grows exponentially with sentence length (\( t^n \)), making it computationally infeasible for real-world sentences.

**Q2: If we have 5 POS tags and a sentence of 8 words, how many tag sequences does brute-force consider?**  
**A:** \( 5^8 = 390,625 \) sequences.

**Q3: What is the main source of redundancy in the brute-force approach?**  
**A:** Many sequences share the same prefix (e.g., `NOUN-VERB-...`) and the probability of that prefix is recomputed for each full sequence.

---

### 6. Revision Notes

- **Brute-force**: Try every tag sequence, pick the best.
- **Complexity**: \( O(t^n) \) – exponential in sentence length.
- **Problem**: Becomes impossible for sentences longer than ~5 words.
- **Why it fails**: Redundant computation and exponential growth.
- **Solution**: Use a dynamic programming algorithm like **Viterbi** (covered next).

## Topic 5: The Viterbi Algorithm – Optimization & Step-by-Step Walkthrough

### 1. Introduction

**What is the Viterbi Algorithm?**
The Viterbi algorithm is a **dynamic programming** algorithm that efficiently finds the most probable sequence of hidden states (POS tags) given a sequence of observations (words). Instead of trying every possible combination like brute-force, Viterbi **systematically keeps only the best path** to each state at each step.

**Why is it important?**
It solves the exponential explosion problem of brute-force by reducing the time complexity from \( O(t^n) \) to \( O(t^2 \times n) \), where:
- \( t \) = number of POS tags
- \( n \) = number of words in the sentence

This makes POS tagging **practical** for sentences of any reasonable length.

**Real-life use:**
Viterbi is used not only in POS tagging but also in:
- Speech recognition (finding the most likely sequence of phonemes).
- Bioinformatics (gene prediction).
- Error correction in communication systems.

---

### 2. Detailed Explanation

#### The Core Idea of Dynamic Programming

The key insight is this:
> *If multiple paths reach the same state at the same time step, we only need to keep the path with the highest probability.*

Why? Because future decisions (transitions to next words) depend only on the **current state**, not on how we got there. This is the **Markov property**—the future is independent of the past given the present.

So, at each word position, for each possible tag, we store:
- **The highest probability** of any tag sequence ending with that tag for that word.
- **A pointer** (backpointer) to the previous tag that gave this highest probability.

At the end, we follow the backpointers from the last word back to the start to reconstruct the optimal tag sequence.

---

#### Step-by-Step Walkthrough (From the Transcript)

Let's apply Viterbi to the same sentence: **"Will Google camping"**  
Tags: `{NOUN, VERB, MODAL}`. We use the emission and transition probabilities we calculated earlier.

**Step 0: Initialize (Word 1 – "Will")**

We start from `START`. For each possible tag for the first word, we compute:

\[
\text{score}(\text{tag}) = P(\text{START} \rightarrow \text{tag}) \times P(\text{"Will"} \mid \text{tag})
\]

| Tag | \( P(\text{START} \rightarrow \text{tag}) \) | \( P(\text{"Will"} \mid \text{tag}) \) | Product |
|-----|----------------------------------------------|----------------------------------------|---------|
| NOUN| 0.75 | 0 (emission is 0) | **0** |
| VERB| 0 | 0 | **0** |
| MODAL| 0.25 | 0.5 | **0.125** |

So only `MODAL` survives. We store:
- `score[0][MODAL] = 0.125`
- `backpointer[0][MODAL] = START`

---

**Step 1: Word 2 – "Google"**

For each current tag \( j \), we consider all previous tags \( i \), and compute:

\[
\text{score}[j] = \max_{i} \big( \text{score}[i] \times P(i \rightarrow j) \times P(\text{"Google"} \mid j) \big)
\]

First, emission probabilities for "Google":
- \( P(\text{"Google"} \mid \text{NOUN}) = 0.3 \)
- \( P(\text{"Google"} \mid \text{VERB}) = 0 \)
- \( P(\text{"Google"} \mid \text{MODAL}) = 0 \)

Since VERB and MODAL have zero emission for "Google", only NOUN is possible.

Now compute the best way to reach NOUN from the previous step (only MODAL survived):

\[
\text{score}[\text{NOUN}] = 0.125 \times P(\text{MODAL} \rightarrow \text{NOUN}) \times 0.3
\]

From transition table:
- \( P(\text{MODAL} \rightarrow \text{NOUN}) = 1.0 \)

So:
\[
\text{score}[\text{NOUN}] = 0.125 \times 1.0 \times 0.3 = 0.0375
\]

We store:
- `score[1][NOUN] = 0.0375`
- `backpointer[1][NOUN] = MODAL`

All other tags have score 0.

---

**Step 2: Word 3 – "camping"**

Emission probabilities for "camping":
- \( P(\text{"camping"} \mid \text{NOUN}) = 0.3 \)
- \( P(\text{"camping"} \mid \text{VERB}) = 0 \)
- \( P(\text{"camping"} \mid \text{MODAL}) = 0 \)

Again, only NOUN is possible.

From previous step, only NOUN survived. Compute:

\[
\text{score}[\text{NOUN}] = 0.0375 \times P(\text{NOUN} \rightarrow \text{NOUN}) \times 0.3
\]

From transition table:
- \( P(\text{NOUN} \rightarrow \text{NOUN}) = 0 \)

So:
\[
\text{score}[\text{NOUN}] = 0.0375 \times 0 \times 0.3 = 0
\]

Uh oh! This means the sequence `MODAL → NOUN → NOUN` has zero probability because in our training data, a noun never followed a noun.

---

**Step 3: Handling the Zero – What Happens Next?**

In the transcript, the sequence `Will (MODAL) → Google (NOUN) → camping (NOUN)` gives **zero** because `NOUN → NOUN` transition was 0.

But the **correct** tagging according to the data is actually:
- `Will` → MODAL
- `Google` → NOUN
- `camping` → NOUN

Wait—that gives zero! So how does Viterbi handle this?

**Important insight from the transcript:**  
The transcript uses a **simplified dataset** where `NOUN → NOUN` was indeed 0, so the algorithm would assign **zero probability** to the correct sequence if it followed only the training data. But in practice, we use **smoothing** or a larger dataset so that no transition is ever exactly 0.

However, in this toy example, the Viterbi algorithm would still find the **best non-zero path**. If no path has non-zero probability, that means the model cannot tag the sentence—which is a sign that the training data is too small.

In the transcript's second example, they changed the transition `VERB → NOUN` to 1 and reran the algorithm to demonstrate how Viterbi chooses between competing paths.

---

#### The General Viterbi Recurrence

Formally, for each word position \( k \) and tag \( j \):

\[
\text{score}_k(j) = \max_{i} \big( \text{score}_{k-1}(i) \times P(i \rightarrow j) \times P(w_k \mid j) \big)
\]

\[
\text{backpointer}_k(j) = \arg\max_{i} \big( \text{score}_{k-1}(i) \times P(i \rightarrow j) \times P(w_k \mid j) \big)
\]

At the end, we add the transition to `END`:

\[
\text{final\_score} = \max_{i} \big( \text{score}_n(i) \times P(i \rightarrow \text{END}) \big)
\]

Then we trace back the backpointers to get the best tag sequence.

---

### 3. Key Points

- **Viterbi is dynamic programming:** It keeps only the best path to each state, discarding all others.
- **Time complexity:** \( O(t^2 \times n) \) instead of \( O(t^n) \).
- **Backpointers** allow reconstruction of the optimal sequence.
- The algorithm uses the **Markov property**: only the current tag matters for future decisions.
- If a transition or emission probability is zero, that path is eliminated immediately.

---

### 4. Syntax / Structure (Pseudo-code)

```python
# Input: sentence (list of words), transition_probs, emission_probs, tags
# Output: best tag sequence

# Initialize
scores = [[0] * len(tags) for _ in range(len(sentence))]
backpointers = [[None] * len(tags) for _ in range(len(sentence))]

# Word 1
for j, tag in enumerate(tags):
    scores[0][j] = start_prob[tag] * emission_probs[tag][sentence[0]]
    backpointers[0][j] = "START"

# Words 2 to n
for k in range(1, len(sentence)):
    for j, current_tag in enumerate(tags):
        best_score = -float('inf')
        best_prev = None
        for i, prev_tag in enumerate(tags):
            score = scores[k-1][i] * transition_probs[prev_tag][current_tag] * emission_probs[current_tag][sentence[k]]
            if score > best_score:
                best_score = score
                best_prev = i
        scores[k][j] = best_score
        backpointers[k][j] = best_prev

# End
best_last_tag = argmax(scores[-1][i] * transition_probs[tags[i]]["END"])
best_seq = trace_back(backpointers, best_last_tag)
```

### 5. Output (Step-by-Step for the Second Example)

In the transcript, the second example changed \( P(\text{VERB} \rightarrow \text{NOUN}) = 1.0 \) to make the problem more interesting. The Viterbi algorithm then had to choose between two paths:

- **Path A:** START → MODAL → NOUN → NOUN → END (via NOUN from MODAL)
- **Path B:** START → MODAL → VERB → NOUN → END (via VERB from MODAL)

At the VERB step, Viterbi compared the scores:
- From MODAL → VERB: score was lower
- From MODAL → NOUN → VERB: not possible (transition 0)

After computing both, Viterbi kept only the higher-scoring path at each state, eventually identifying that **START → MODAL → NOUN → VERB → NOUN → END** was the best (or whatever the data supported).

The key is: **Viterbi did not try all 81 combinations; it kept only the best path to each state (at most 3 paths per step).**

### 6. Common Mistakes

- **Forgetting to multiply by emission probabilities:** Both transition and emission matter equally.
- **Using scores without backpointers:** Without backpointers, you can't reconstruct the sequence.
- **Assuming Viterbi always finds a non-zero path:** If the training data is too small, many paths may have zero probability. This is a data problem, not an algorithm problem.
- **Not handling the END transition:** The last word must transition to END. Forgetting this gives incorrect scores.
