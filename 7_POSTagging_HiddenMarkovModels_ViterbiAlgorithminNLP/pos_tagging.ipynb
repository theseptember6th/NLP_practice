## Topic 2: Hands-on POS Tagging using spaCy (Code, Attributes & Visualization)

### 1. Introduction

**What is spaCy?**
spaCy is a popular, open-source Python library designed specifically for advanced NLP tasks. It is **production-ready** and comes with pre-trained models that can perform POS tagging, tokenization, named entity recognition, and more—right out of the box.

**Why use spaCy for POS tagging?**
- It is **fast** and **efficient**.
- It provides **two levels** of POS tags: coarse-grained (simplified) and fine-grained (detailed).
- It includes a built-in visualizer (`displacy`) to see the tags graphically.
- You don't need to train a model; spaCy's pre-trained model does all the heavy lifting.

**Real-life use:**
If you are building a chatbot, a text summarizer, or any language tool, you can use spaCy to quickly get POS tags without writing complex algorithms yourself.

---

### 2. Detailed Explanation

#### Step-by-step workflow

1. **Install spaCy** (if not already installed).
2. **Import the library** and load the English language model.
3. **Create a document object** from your input sentence.
4. **Access tags** using attributes like `.pos_` (coarse) and `.tag_` (fine-grained).
5. **Use the `spacy.explain()`** function to get the full meaning of a tag.
6. **Visualize** the tags using `displacy`.

---

#### Understanding the Attributes

When you process a text with spaCy, it returns a `Doc` object. Each token (word) in this object has several important attributes:

| Attribute | What it gives | Example |
|-----------|---------------|---------|
| `.text` | The actual word | `"Google"` |
| `.pos_` | Coarse-grained POS tag (short, general) | `"PROPN"` (proper noun) |
| `.tag_` | Fine-grained POS tag (more specific) | `"NNP"` (proper noun, singular) |
| `.pos` / `.tag` | Integer IDs (not used for beginners) | — |

**Coarse vs. Fine-grained:**
- **Coarse (`.pos_`):** Broad category (e.g., `NOUN`, `VERB`, `ADJ`). Good for general understanding.
- **Fine-grained (`.tag_`):** Detailed tag (e.g., `NNP` for proper noun, `VBD` for past tense verb). Good for precise grammar analysis.

**Using `spacy.explain()`**
If you don't understand a tag, you can pass it to `spacy.explain()` to get a plain-English description.

Example:
```python
spacy.explain("PRP")   # Output: "pronoun, personal"
spacy.explain("VBD")   # Output: "verb, past tense"
```

### 3. Key Points

- spaCy does POS tagging automatically behind the scenes.

- Fine-grained tags give more grammatical detail than coarse ones.

### 4. Syntax / Structure (Code Examples)

Below are all the code snippets from the transcript, explained line by line.

#### Example 1: Basic setup and loading model

```python
# Install spaCy (only once)
# !pip install spacy

# Import the library
import spacy

# Load the English language model (small version)
nlp = spacy.load("en_core_web_sm")
```

**Explanation:**
- `import spacy` brings the library into your environment.
- `spacy.load("en_core_web_sm")` loads a small, efficient English model. This model contains the rules and data needed for tagging. We store it in the variable `nlp` (common convention).

---

#### Example 2: Processing a sentence

```python
# Create a document object from a sentence
doc = nlp("I will Google about Facebook")

# Print the entire sentence
print(doc.text)
```

**Explanation:**
- `nlp("I will Google about Facebook")` processes the sentence and returns a `Doc` object named `doc`.
- `doc.text` returns the original text.

---

#### Example 3: Indexing into the document

```python
# Access tokens like a list
print(doc[0])    # Output: "I"
print(doc[1])    # Output: "will"
print(doc[-1])   # Output: "Facebook"
```

**Explanation:**
- The `Doc` object supports indexing like a list. `doc[0]` gives the first token, `doc[1]` gives the second, and `doc[-1]` gives the last token.

---

#### Example 4: Getting POS tags for a single token

```python
# Coarse-grained tag
print(doc[0].pos_)   # Output: "PRON" (pronoun)

# Fine-grained tag
print(doc[0].tag_)   # Output: "PRP" (personal pronoun)

# Explanation of the tag
print(spacy.explain(doc[0].tag_))   # Output: "pronoun, personal"
```

**Explanation:**
- `.pos_` gives the coarse tag.
- `.tag_` gives the fine-grained tag.
- `spacy.explain()` translates the tag code into a readable description.

---

#### Example 5: Iterating over all tokens

```python
for token in doc:
    print(f"{token.text:10} --> {token.pos_:6} --> {token.tag_:6} --> {spacy.explain(token.tag_)}")
```

**Output (explained):**
```
I          --> PRON   --> PRP    --> pronoun, personal
will       --> AUX    --> MD     --> verb, modal auxiliary
Google     --> PROPN  --> NNP    --> noun, proper singular
about      --> ADP    --> IN     --> conjunction, subordinating or preposition
Facebook   --> PROPN  --> NNP    --> noun, proper singular
```

**Explanation:**
- The `for` loop goes through each token.
- `token.text` is the word.
- `token.pos_` is the coarse tag.
- `token.tag_` is the fine-grained tag.
- `spacy.explain(token.tag_)` gives the meaning.

---

#### Example 6: Word Sense Disambiguation with spaCy

```python
# First sentence
doc1 = nlp("I left the room")
print(doc1[1].pos_)   # Output: "VERB" (left as an action)

# Second sentence
doc2 = nlp("the left side of the room")
print(doc2[1].pos_)   # Output: "ADJ" (left as a descriptor)
```

**Explanation:**
- The same word "left" receives different POS tags based on context, proving spaCy handles ambiguity.

---

#### Example 7: Visualizing POS tags with displacy

```python
from spacy import displacy

# New sentence
doc = nlp("The quick brown fox jumps over the lazy dog")

# Render the visualization in Jupyter Notebook
displacy.render(doc, style="dep", jupyter=True)
```

**Explanation:**
- `from spacy import displacy` imports the visualizer.
- `displacy.render(doc, style="dep", jupyter=True)` draws a dependency graph showing POS tags and relationships between words.

---

#### Example 8: Customizing the visualization

```python
# Custom options dictionary
options = {"distance": 80, "bg": "#ffffff", "color": "blue"}

# Render with custom options
displacy.render(doc, style="dep", options=options, jupyter=True)
```

**Explanation:**
- You can pass an `options` dictionary to control appearance:
  - `distance`: spacing between words.
  - `bg`: background color.
  - `color`: color of the arrows/connections.

### 6. Common Mistakes

- **Forgetting to load the model:** You must run `nlp = spacy.load("en_core_web_sm")` before processing text.
- **Using `.pos` instead of `.pos_`:** The underscore (`_`) returns the string version. Without it, you get an integer ID, which is not human-readable.
- **Not installing the model:** `spacy.load()` will fail if the model is not downloaded. Install it with `python -m spacy download en_core_web_sm`.
- **Confusing coarse and fine tags:** Remember: `.pos_` is general, `.tag_` is specific.
