# Topic 1: Project Overview & Problem Statement

## 1. Introduction

This project focuses on **Duplicate Question Pairs** – a classic Natural Language Processing (NLP) problem. The goal is to determine whether two given questions are duplicates (have the same meaning) or not.

**Real-life use:** Websites like Quora, Stack Overflow, and other Q&A platforms face the issue where users ask the same question using different wording. This creates redundancy, splits answers across multiple pages, and degrades user experience. By automatically identifying duplicate questions, platforms can:
- Merge similar questions
- Redirect users to existing answers
- Improve content organization
- Save time for both answerers and askers

**Dataset:** The project uses Quora's dataset from a competition conducted by Quora and Kaggle. It contains approximately **400,000 rows**, each with:
- `id`: Unique identifier
- `qid1`: Question 1 ID
- `qid2`: Question 2 ID  
- `question1`: The first question text
- `question2`: The second question text
- `is_duplicate`: Label (0 = not duplicate, 1 = duplicate)

---

## 2. Detailed Explanation

### What is a Duplicate Question?

Two questions are considered duplicates if they **ask the same thing**, even if the wording differs.

**Example of duplicates:**
- Question 1: "Who is the Prime Minister of India?"
- Question 2: "Who is the current Prime Minister of India?"

Both ask for the same information – the words are slightly different but the meaning is identical.

**Example of non-duplicates:**
- Question 1: "Who is the Prime Minister of India?"
- Question 2: "Who is the Prime Minister of Pakistan?"

These are different questions entirely (different countries), so they are **not duplicates**.

### Why is this a Classification Problem?

This is a **binary classification** problem:
- **Class 1 (Duplicate)** → Questions have the same meaning
- **Class 0 (Not Duplicate)** → Questions have different meanings

We feed two questions into a machine learning model, and the model outputs a prediction: 0 or 1.

This served as the baseline to measure improvements against.

# Topic 3: Basic Feature Engineering (7 Features)

## 1. Introduction

**Feature engineering** is the process of creating new features (columns) from existing data to help machine learning models make better predictions. Raw text isn't directly usable by models – we need to convert it into meaningful numerical features.

In this project, the instructor created **7 basic features** from the question pairs. These features capture simple patterns that might indicate whether two questions are duplicates.

**Why feature engineering matters:**
- Raw text alone gave ~73.74% accuracy
- Adding these 7 features improved accuracy to **~76.80%**
- Even simple features provided a **~3% improvement**

---

## 2. Detailed Explanation

### The 7 Basic Features

| # | Feature Name | Description | How to Calculate |
|---|--------------|-------------|------------------|
| 1 | `q1_len` | Length of question 1 (number of characters) | `len(question1)` |
| 2 | `q2_len` | Length of question 2 (number of characters) | `len(question2)` |
| 3 | `q1_num_words` | Number of words in question 1 | Count words (split by space) |
| 4 | `q2_num_words` | Number of words in question 2 | Count words (split by space) |
| 5 | `common_words` | Number of common words between both questions | Intersection of word sets |
| 6 | `total_words` | Total number of unique words across both questions | Union of word sets |
| 7 | `word_share` | Proportion of common words relative to total words | `common_words / total_words` |

---

### Feature 1 & 2: Question Length (Character Count)

**What it captures:** Longer questions may contain more context, making them less likely to be duplicates of shorter questions. Or, both questions might be similarly long if they ask the same thing.

**Calculation:**
```python
q1_len = len(question1)      # Number of characters in question 1
q2_len = len(question2)      # Number of characters in question 2
```

**Example:**
- Question 1: "Who is the Prime Minister of India?" → 34 characters
- Question 2: "Who is the current Prime Minister of India?" → 41 characters

**Observation from data:**
- Minimum length: ~6 characters
- Maximum length: ~1,000+ characters
- Average length: ~50-60 characters
- Most questions fall in the 25-70 character range (seen in histograms)

---

### Feature 3 & 4: Word Count

**What it captures:** Similar to length, but counts words instead of characters. Questions with very different word counts might be less likely to be duplicates.

**Calculation:**
```python
q1_num_words = len(question1.split())      # Split by space and count
q2_num_words = len(question2.split())
```

**Example:**
- Question 1: "Who is the Prime Minister of India?" → 7 words
- Question 2: "Who is the current Prime Minister of India?" → 8 words

**Observation from data:**
- Minimum words: 2-3
- Maximum words: 172+
- Average: ~15-20 words per question

---

### Feature 5: Common Words

**What it captures:** The number of words that appear in **both** questions. If two questions share many words, they're more likely to be duplicates.

**Calculation:**
```python
# Convert questions to sets of words
words1 = set(question1.split())
words2 = set(question2.split())

# Find intersection (common words)
common_words = len(words1.intersection(words2))
```

**Example:**
- Q1: "Who is the Prime Minister of India?" → {Who, is, the, Prime, Minister, of, India}
- Q2: "Who is the current Prime Minister of India?" → {Who, is, the, current, Prime, Minister, of, India}
- Common words: {Who, is, the, Prime, Minister, of, India} → **7 common words**

**Key insight from visualization:**
- Questions with **< 4 common words** are more likely to be **non-duplicate**
- Questions with **≥ 4 common words** are more likely to be **duplicate**
- This feature alone showed clear class separation

---

### Feature 6: Total Words (Unique Words Across Both Questions)

**What it captures:** The total number of **unique** words across both questions (union). This helps normalize the common word count.

**Calculation:**
```python
total_words = len(words1.union(words2))
```

**Example:**
- Q1 words: {Who, is, the, Prime, Minister, of, India}
- Q2 words: {Who, is, the, current, Prime, Minister, of, India}
- Union: {Who, is, the, Prime, Minister, of, India, current} → **8 unique words**

---

### Feature 7: Word Share

**What it captures:** The proportion of words that are common between the two questions. This normalizes for question length.

**Calculation:**
```python
word_share = common_words / total_words
```

**Example from above:**
- `common_words` = 7
- `total_words` = 8
- `word_share` = 7/8 = 0.875 (87.5% overlap)

**Key insight from visualization:**
- Questions with **word_share < 0.2** are more likely to be **non-duplicate**
- Questions with **word_share ≥ 0.2** are more likely to be **duplicate**
- This provides a clear threshold for classification

---

### Feature Creation Process

The instructor used this general approach:

1. Take the raw dataset (question1, question2)
2. For each row, calculate the 7 features
3. Add these as new columns to the dataset
4. Combine with TF-IDF features (3,000 per question)
5. Train Random Forest model

**Result:** Accuracy improved from **73.74%** to **76.80%** – a ~3% gain!

---

## 4. Code Example (Feature Engineering)

```python
# Creating the 7 basic features
def create_basic_features(df):
    # Question length (character count)
    df['q1_len'] = df['question1'].str.len()
    df['q2_len'] = df['question2'].str.len()
    
    # Word count
    df['q1_num_words'] = df['question1'].str.split().str.len()
    df['q2_num_words'] = df['question2'].str.split().str.len()
    
    # Common words
    df['common_words'] = df.apply(
        lambda x: len(set(x['question1'].split()).intersection(
                     set(x['question2'].split()))), axis=1
    )
    
    # Total unique words
    df['total_words'] = df.apply(
        lambda x: len(set(x['question1'].split()).union(
                     set(x['question2'].split()))), axis=1
    )
    
    # Word share (proportion)
    df['word_share'] = df['common_words'] / df['total_words']
    
    return df
```

**Explanation:**
- `.str.len()` → Counts characters
- `.str.split().str.len()` → Splits by space and counts words
- `set()` → Converts to set (removes duplicates)
- `.intersection()` → Finds common words
- `.union()` → Finds all unique words
- Division creates proportion

# Topic 4: Data Preprocessing & Text Cleaning

## 1. Introduction

**Text preprocessing** is the process of cleaning and normalizing raw text data before feeding it into a machine learning model. Raw text contains noise like:
- Uppercase/lowercase variations
- Special characters and punctuation
- Numbers and symbols
- Contractions (e.g., "don't" → "do not")
- Stop words (common words like "the", "is", "and")

In this project, preprocessing was identified as a **critical step** for improving model performance. Without proper preprocessing, even advanced features may not help much.

---

## 2. Detailed Explanation

### Why Preprocessing Matters

The instructor observed that the **baseline model (73.74%)** and even the **basic features model (76.80%)** still had room for improvement. The next logical step was to clean the text data so that feature extraction (like TF-IDF and word counts) becomes more meaningful.

**Key insight:** Two questions might be duplicates but have different:
- Capitalization ("India" vs "india")
- Punctuation ("Prime Minister?" vs "Prime Minister")
- Special characters ("₹100" vs "100 rupees")
- Contractions ("don't" vs "do not")
- Abbreviations ("Dr." vs "Doctor")

By standardizing all of these, the model can focus on **meaning** rather than **formatting**.

---

The instructor created a function that applies all preprocessing steps in sequence. Here's the full pipeline:

| Step | Operation | Description |

| 1 | Lowercasing | Convert all text to lowercase |
| 2 | Whitespace stripping | Remove leading/trailing spaces |
| 3 | Special character replacement | Convert symbols to text equivalents |
| 4 | Math expression removal | Remove mathematical expressions that don't add meaning |
| 5 | Number/symbol removal | Remove numbers and special symbols |

### Step 6: Contraction Expansion

**What it does:** Expands contractions to their full forms.

**Why:** Prevents "don't" and "do not" from being treated as different words.

**Common contractions:**
| Contraction | Expanded Form |
|-------------|---------------|
| don't | do not |

```python
# Dictionary of contractions
contraction_dict = {
    "don't": "do not",
    "can't": "cannot",
    "won't": "will not",
    # ... many more
}

# Apply expansion
for contraction, expanded in contraction_dict.items():
    question = question.replace(contraction, expanded)
```

## 2. Detailed Explanation

### Category 1: Token Features (6 Features)

**Token** = An individual word in a sentence. Token features compare words between the two questions.

#### Prerequisite: Understanding Stop Words

Before understanding token features, we need to understand the difference between:

| Term | Definition | Example |
|------|------------|---------|
| **Token** | Any word in a sentence | "the", "cat", "sat" |
| **Stop Word** | Common words that carry little meaning | "the", "is", "and" |
| **Normal Word** | Meaningful content words | "prime", "minister", "India" |

**Example:**
- Sentence: "The Prime Minister of India"
- Tokens: ["The", "Prime", "Minister", "of", "India"]
- Stop words: ["The", "of"]
- Normal words: ["Prime", "Minister", "India"]

---

#### Token Feature 1: Number of Common Stop Words (`cwc_min`)

**What it measures:** The number of common stop words between both questions, divided by the minimum number of stop words in either question.

**Formula:**
```
cwc_min = (common_stop_words) / min(stop_words_in_q1, stop_words_in_q2)
```

**Why useful:** Duplicate questions often share the same stop words structure. If both questions have similar stop word patterns, they might be duplicates.

**Example:**
- Q1: "Who is the Prime Minister of India?" → Stop words: ["who", "is", "the", "of"] → 4 stop words
- Q2: "Who is the current Prime Minister of India?" → Stop words: ["who", "is", "the", "of"] → 4 stop words
- Common stop words: ["who", "is", "the", "of"] → 4 common
- `cwc_min` = 4 / min(4, 4) = 4/4 = 1.0 (high overlap)

---

#### Token Feature 2: Number of Common Stop Words with Max Denominator (`cwc_max`)

**Formula:**
```
cwc_max = (common_stop_words) / max(stop_words_in_q1, stop_words_in_q2)
```

**Why useful:** Normalizes by the larger question length instead of the smaller. This penalizes cases where one question has many extra stop words.

**Example from above:**
- `cwc_max` = 4 / max(4, 4) = 4/4 = 1.0

**If questions differ:**
- Q1: 10 stop words, Q2: 4 stop words, common: 4
- `cwc_min` = 4/4 = 1.0 (appears perfect)
- `cwc_max` = 4/10 = 0.4 (reveals the difference)

---

#### Token Feature 3: Number of Common Normal Words with Min Denominator (`ctc_min`)

**What it measures:** Same as `cwc_min`, but for **normal words** (non-stop words).

**Formula:**
```
ctc_min = (common_normal_words) / min(normal_words_in_q1, normal_words_in_q2)
```

**Why useful:** Normal words carry the actual meaning of the question. Duplicate questions should share many normal words.

**Example:**
- Q1: "prime minister india" → Normal words: ["prime", "minister", "india"] → 3
- Q2: "current prime minister india" → Normal words: ["current", "prime", "minister", "india"] → 4
- Common normal words: ["prime", "minister", "india"] → 3
- `ctc_min` = 3 / min(3, 4) = 3/3 = 1.0 (perfect overlap)

---

#### Token Feature 4: Number of Common Normal Words with Max Denominator (`ctc_max`)

**Formula:**
```
ctc_max = (common_normal_words) / max(normal_words_in_q1, normal_words_in_q2)
```

**Example from above:**
- `ctc_max` = 3 / max(3, 4) = 3/4 = 0.75

---

#### Token Feature 5: Last Word Equality (`last_word`)

**What it measures:** Whether the last word of both questions is the same.

**Calculation:**
```python
last_word = 1 if q1.split()[-1] == q2.split()[-1] else 0
```

**Why useful:** The last word often indicates the core subject of the question.
- "What is India?" → Last word: "India"
- "Tell me about India" → Last word: "India"
- If last words match, there's higher chance of being duplicates

**Key insight from visualization:**
- If last words **don't match** → Very high probability of **non-duplicate**
- If last words **match** → Mixed, but often indicates duplicates

---

#### Token Feature 6: First Word Equality (`first_word`)

**What it measures:** Whether the first word of both questions is the same.

**Calculation:**
```python
first_word = 1 if q1.split()[0] == q2.split()[0] else 0
```

**Why useful:** The first word often indicates the question type (What, Who, How, Why, When, Where).
- If both start with "What" → likely asking about the same type of information
- If one starts with "Who" and the other with "What" → likely different types of questions

**Example:**
- Q1: "Who is the President?" → First word: "Who"
- Q2: "Who is the current President?" → First word: "Who" → Match! (likely duplicate)
- Q3: "What is the President?" → First word: "What" → No match!

### Category 2: Length-Based Features (3 Features)

These features analyze the length of questions in terms of **tokens** (words).

#### Length Feature 1: Mean Length (`mean_len`)

**What it measures:** The average number of tokens (words) across both questions.

**Formula:**
```
mean_len = (tokens_in_q1 + tokens_in_q2) / 2
```

**Why useful:** Questions that are similar in length might be asking the same thing.

**Example:**
- Q1: 7 tokens, Q2: 8 tokens → mean_len = 7.5

---

#### Length Feature 2: Absolute Difference in Length (`len_diff`)

**What it measures:** The absolute difference in token counts between both questions.

**Formula:**
```
len_diff = |tokens_in_q1 - tokens_in_q2|
```

**Why useful:** Large difference in length suggests different complexity, making duplication less likely.

**Example:**
- Q1: 7 tokens, Q2: 8 tokens → len_diff = |7 - 8| = 1 (small → likely duplicate)
- Q1: 7 tokens, Q2: 25 tokens → len_diff = 18 (large → likely non-duplicate)

---

#### Length Feature 3: Longest-to-Shortest Ratio (`len_ratio`)

**What it measures:** The ratio of the longer question's length to the shorter question's length.

**Formula:**
```
len_ratio = max(tokens_q1, tokens_q2) / min(tokens_q1, tokens_q2)
```

**Why useful:** A ratio close to 1 means questions are similar in length. A ratio far from 1 means one question is much longer.

**Example:**
- Q1: 7 tokens, Q2: 8 tokens → len_ratio = 8/7 ≈ 1.14 (close to 1 → likely duplicate)
- Q1: 7 tokens, Q2: 25 tokens → len_ratio = 25/7 ≈ 3.57 (far from 1 → likely non-duplicate)

## 4. Code Examples

### Token Features Implementation

```python
def create_token_features(df):
    """Create 6 token-based features"""
    
    # Extract stop words from NLTK
    from nltk.corpus import stopwords
    stop_words = set(stopwords.words('english'))
    
    def get_stop_words(text):
        return [w for w in text.split() if w in stop_words]
    
    def get_normal_words(text):
        return [w for w in text.split() if w not in stop_words]
    
    # Calculate for each row
    results = []
    for idx, row in df.iterrows():
        q1_stop = get_stop_words(row['q1_clean'])
        q2_stop = get_stop_words(row['q2_clean'])
        q1_normal = get_normal_words(row['q1_clean'])
        q2_normal = get_normal_words(row['q2_clean'])
        
        # Common words
        common_stop = len(set(q1_stop) & set(q2_stop))
        common_normal = len(set(q1_normal) & set(q2_normal))
        
        # cwc_min and cwc_max
        min_stop = min(len(q1_stop), len(q2_stop))
        max_stop = max(len(q1_stop), len(q2_stop))
        cwc_min = common_stop / (min_stop + 1e-5)  # Avoid division by zero
        cwc_max = common_stop / (max_stop + 1e-5)
        
        # ctc_min and ctc_max
        min_normal = min(len(q1_normal), len(q2_normal))
        max_normal = max(len(q1_normal), len(q2_normal))
        ctc_min = common_normal / (min_normal + 1e-5)
        ctc_max = common_normal / (max_normal + 1e-5)
        
        # Last and first word equality
        last_word = 1 if q1_clean.split()[-1] == q2_clean.split()[-1] else 0
        first_word = 1 if q1_clean.split()[0] == q2_clean.split()[0] else 0
        
        results.append([cwc_min, cwc_max, ctc_min, ctc_max, last_word, first_word])
    
    return pd.DataFrame(results, columns=['cwc_min', 'cwc_max', 'ctc_min', 'ctc_max', 
                                          'last_word', 'first_word'])
```

### Length Features Implementation

```python
def create_length_features(df):
    """Create 3 length-based features"""
    
    results = []
    for idx, row in df.iterrows():
        tokens1 = len(row['q1_clean'].split())
        tokens2 = len(row['q2_clean'].split())
        
        mean_len = (tokens1 + tokens2) / 2
        len_diff = abs(tokens1 - tokens2)
        len_ratio = max(tokens1, tokens2) / (min(tokens1, tokens2) + 1e-5)
        
        results.append([mean_len, len_diff, len_ratio])
    
    return pd.DataFrame(results, columns=['mean_len', 'len_diff', 'len_ratio'])
```

### Fuzzy Features Implementation

```python
from fuzzywuzzy import fuzz

def create_fuzzy_features(df):
    """Create 4 fuzzy matching features"""
    
    results = []
    for idx, row in df.iterrows():
        q1 = row['q1_clean']
        q2 = row['q2_clean']
        
        fuzz_ratio = fuzz.ratio(q1, q2)
        fuzz_partial = fuzz.partial_ratio(q1, q2)
        token_sort = fuzz.token_sort_ratio(q1, q2)
        token_set = fuzz.token_set_ratio(q1, q2)
        
        results.append([fuzz_ratio, fuzz_partial, token_sort, token_set])
    
    return pd.DataFrame(results, columns=['fuzz_ratio', 'fuzz_partial_ratio', 
                                          'token_sort_ratio', 'token_set_ratio'])
```

**Why accuracy alone isn't enough:**

#### Confusion Matrix

A confusion matrix shows:
- **True Positives (TP):** Correctly predicted duplicates
- **True Negatives (TN):** Correctly predicted non-duplicates
- **False Positives (FP):** Predicted duplicate, but actually non-duplicate (Type I error)
- **False Negatives (FN):** Predicted non-duplicate, but actually duplicate (Type II error)

```
                Predicted
              Duplicate  Non-Duplicate
Actual  Duplicate    TP          FN
        Non-Dup      FP          TN
```

**Critical Insight – Which error is worse?**

#### Key Functions in `helper.py`

```python
def query_point_creator(q1, q2):
    """
    Creates the full feature vector for two input questions.
    Returns: 1 × 6022 vector ready for prediction
    """
    # Step 1: Preprocess both questions
    q1_clean = preprocess_question(q1)
    q2_clean = preprocess_question(q2)
    
    # Step 2: Create 7 basic features
    basic_features = create_basic_features(q1_clean, q2_clean)
    
    # Step 3: Create 15 advanced features
    token_features = create_token_features(q1_clean, q2_clean)
    length_features = create_length_features(q1_clean, q2_clean)
    fuzzy_features = create_fuzzy_features(q1_clean, q2_clean)
    
    # Step 4: Apply TF-IDF vectorizer
    q1_tfidf = vectorizer.transform([q1_clean])
    q2_tfidf = vectorizer.transform([q2_clean])
    
    # Step 5: Combine all features
    features = np.concatenate([
        basic_features,
        token_features,
        length_features,
        fuzzy_features,
        q1_tfidf.toarray().flatten(),
        q2_tfidf.toarray().flatten()
    ])
    
    return features.reshape(1, -1)
```

#### Main Application in `app.py`

```python
from flask import Flask, request, render_template
import pickle
import numpy as np
from helper import query_point_creator

app = Flask(__name__)

# Load trained model
model = pickle.load(open('model.pkl', 'rb'))

@app.route('/')
def home():
    return render_template('index.html')

@app.route('/predict', methods=['POST'])
def predict():
    # Get questions from form
    q1 = request.form['question1']
    q2 = request.form['question2']
    
    # Create feature vector
    features = query_point_creator(q1, q2)
    
    # Predict
    prediction = model.predict(features)[0]
    
    # Get probability
    probability = model.predict_proba(features)[0][1]  # Probability of being duplicate
    
    # Interpret result
    if prediction == 1:
        result = "Duplicate Questions"
        message = f"These questions are likely duplicates (confidence: {probability:.2%})"
    else:
        result = "Not Duplicate"
        message = f"These questions are not duplicates (confidence: {1-probability:.2%})"
    
    return render_template('index.html', 
                         prediction=result, 
                         message=message,
                         q1=q1,
                         q2=q2)

if __name__ == '__main__':
    app.run(debug=True)
```

## 2. Detailed Explanation

### Observation 1: The Importance of Probability Thresholds

By default, machine learning models use a **0.5 threshold** for classification:
- Probability ≥ 0.5 → Class 1 (Duplicate)
- Probability < 0.5 → Class 0 (Not Duplicate)

**The instructor's observation:** The model provides probability scores between 0 and 1. Instead of using the default 0.5 threshold, we can **adjust the threshold** to prioritize reducing False Positives.

**Example:**
```
If probability = 0.65:
- With threshold 0.5 → Predict Duplicate (1)
- With threshold 0.7 → Predict Not Duplicate (0)
```

**Why adjust thresholds?**
- **Higher threshold** (e.g., 0.7) → Fewer False Positives (safer – only predict duplicate when very confident)
- **Lower threshold** (e.g., 0.3) → More True Positives but more False Positives

**Recommendation:** For this project, use a **higher threshold** (e.g., 0.6 or 0.7) to avoid incorrectly merging different questions.
